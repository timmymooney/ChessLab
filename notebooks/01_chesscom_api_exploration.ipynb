# ChessLab API Exploration

This notebook is used to explore the Chess.com public API and verify that
ChessLab can:

1. Retrieve a player's available game archives.
2. Download games for a specific month.
3. Inspect the returned PGN data.

This is exploratory work. Stable functionality will eventually live in
`src/chesslab/`.

```python
# import functions from 'src/chesslab/api/chesscom.py'
from chesslab.api.chesscom import (
    get_game_archives,
    download_monthly_games,
)
```

## 1. Choose a player/username

For this initial exploration, we'll use my own chess data (as shocking as it may be).

Later, this will become user input in the ChessLab application.

```python
# using my own username and printing expected output
username = "tjmoonz98"
print(f"Exploring games for: {username}")
```

## 2. Retrieve available game archives

Chess.com organises a player's games into monthly archives.

Let's see which months are available for this player.

```python
# define archives with the get_game_archives function from ChessLab
archives = get_game_archives(username)

# print number of available data to use
print(f"Number of available monthly archives: {len(archives)}")
print()

# list the last 3 months of data for Hikaru
for archive in archives[-3:]:
    print(archive)
```

## 3. Select data for a given month

Let's inspect one of the available monthly archives.

We'll use the most recent archive returned by the API.

```python
# define latest archive and print
latest_archive = archives[-1]
print(latest_archive)
```

## 4. Download the month's games

Our `download_monthly_games()` function expects a username, year and month.

For now, we'll extract the year and month from the archive URL manually.

```python
# split the URL into parts
parts = latest_archive.rstrip("/").split("/")
# extract and print the year and month from the URL
year = int(parts[-2])
month = int(parts[-1])
print(f"Year: {year}")
print(f"Month: {month}")
```

```python
# define a Portable Game Notation (PGN) object using the donwload_monthly_games object from ChessLab
pgn = download_monthly_games(
    username=username,
    year=year,
    month=month,
)
# print the number of characters downloaded from the PGN data
print(f"Downloaded {len(pgn):,} characters of PGN data.")
```

## 5. Inspect the raw PGN

PGN (Portable Game Notation) is the standard text format used to represent
chess games.

Let's look at the beginning of the response.

```python
# print the first 750 characters of the PGN object created
print(pgn[:750])
```

## 6. Basic observations

At this stage we're only interested in confirming that the API is returning
real game data.

We will eventually use the `python-chess` library to properly parse these games rather
than working with the raw PGN text.

```python
# print objects to show what we can define at this stage
print("Username:", username)
print("Latest archive:", latest_archive)
print("PGN characters:", len(pgn))
```

## What we've learned

The initial API exploration confirms that ChessLab can:

- Identify a player's available game archives
- Select any given monthly archive
- Download the corresponding PGN data
- Access the raw chess game data

The next step is to parse this PGN into individual games using
`python-chess`.

That will allow ChessLab to work with individual moves, positions,
players, results and eventually engine evaluations.
